```python
import objaverse
import random

random.seed(42)
MAX_VERTEX_COUNT = 5000

def download_and_inspect_subset(num_objects=10, max_vertex_count=MAX_VERTEX_COUNT, download=True):
    print("Fetching UIDs...")
    # This fetches a lightweight list of all ~800k object UIDs
    all_uids = objaverse.load_uids()
    
    # Grab a small subset
    subset_uids = all_uids[:1000]
    print(f"Selected {len(subset_uids)} UIDs to inspect and download.\n")

    # 1. Inspecting the objects (Metadata)
    print("Downloading annotations/metadata...")
    annotations = objaverse.load_annotations(uids=subset_uids)
    
    print("\n--- Object Metadata Preview ---")


    sample_uids = []
    while len(sample_uids) < num_objects:
        uid = random.choice(subset_uids)
        
        # Extract a few interesting details from the metadata dictionary
        metadata = annotations[uid]
        name = metadata.get('name', 'Unknown')
        categories = [cat.get('name') for cat in metadata.get('categories', [])]
        vertex_count = metadata.get('vertexCount')

        if categories and uid not in sample_uids and vertex_count < MAX_VERTEX_COUNT:
            sample_uids.append(uid)

            print(f"UID: {uid}")
            print(f"Name: {name}")
            print(f"Categories: {categories}")
            print(f"Vertex COunt: {vertex_count}")
            print("-" * 30)
        else:
            continue
    
    if download:
        # 2. Downloading the actual 3D objects (.glb files)
        print("\nDownloading 3D models...")
        # This function automatically downloads the GLB files and caches them locally
        # It returns a dictionary mapping the UID to the local file path
        objects = objaverse.load_objects(uids=subset_uids)
        
        print("\n--- Download Complete ---")
        for uid, local_path in objects.items():
            print(f"Model saved to: {local_path}")
    return sample_uids
```

```python
download_and_inspect_subset(5, download=False)
```

```python
from pathlib import Path

def list_downloaded_models():
    # .expanduser() automatically converts "~" to your actual home directory path
    objaverse_dir = Path("~/.objaverse").expanduser()
    
    # .rglob() recursively searches all subdirectories for matching files
    glb_files = list(objaverse_dir.rglob("*.glb"))
    
    print(f"Found {len(glb_files)} downloaded .glb files in {objaverse_dir}:\n")
    
    for file_path in glb_files:
        # .resolve() ensures we print the absolute, full path
        print(file_path.resolve())

    return glb_files

downloaded_paths = list_downloaded_models()
```
